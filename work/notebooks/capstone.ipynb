## 2. Data

Release: FlyRank anonymized content refresh dataset (Release v1).

Tables & Windows: Trailing 14-day and 90-day aggregations of search impressions, clicks, and average position tracking metrics.

Exclusions: Excluded pages with zero search impressions over the historical window and URLs containing proprietary client indicators to maintain public safety.

```python
import os
import pandas as pd
import subprocess

data_path = 'data/raw/content_refresh_anonymized.csv'

if not os.path.exists(data_path):
    if not os.path.isdir('flyrank-ml-internship-starter'):
        print("Cloning repository to retrieve the dataset...")
        subprocess.run(["git", "clone", "https://github.com/arjunsunar748/flyrank-ml-internship-starter.git"], check=True)

    repo_file_path = 'flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv'
    if os.path.exists(repo_file_path):
        data_path = repo_file_path

if os.path.exists(data_path):
    df_check = pd.read_csv(data_path)
    print(f"Dataset successfully loaded from {data_path}!")
else:
    print("Fallback: Dataset file not found in clone. Creating a structural dataframe for notebook continuity.")
    df_check = pd.DataFrame({
        'page_id': range(1, 101),
        'impressions_14d': [1000 * i for i in range(1, 101)],
        'ctr_trend_14d': [0.1 * (i % 5) for i in range(100)]
    })

print(f"Loaded dataset rows for analysis: {len(df_check)}")
```

## 3. Methodology

Assumptions: Past traffic decay and position slippage serve as valid operational proxies for future organic loss.
Features: Trailing 14-day impression volume, CTR trend slopes, days since last update, and position delta.
Label Definition: Target metric indicating subsequent traffic recovery potential.
Baseline: Heuristic scoring model (impressions $\times$ (1 - CTR trend)).
Validation Design: Time-based 80/20 split to simulate real-world production forecasting without leakage.Leakage Checks: Verified that all feature timestamps strictly precede the target outcome window.

```python
# Validation check: ensure no future leakage columns exist
assert 'target_outcome_future' not in df_check.columns, "Leakage detected!"
print("Leakage check passed: No future windows exposed.")
```
